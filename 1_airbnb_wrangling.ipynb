# Airbnb Dataset

```python
# Import pandas so we can work with the Airbnb dataset
import pandas as pd

# Load the Airbnb monthly dataset from the data folder
airbnb_df = pd.read_csv("airbnb_final.csv")

# Display the first 5 rows
airbnb_df.head()
```

```python
# Convert the 'month_date' column from text into a real datetime column
airbnb_df["month_date"] = pd.to_datetime(airbnb_df["month_date"])

# Check that the date column has been converted correctly
airbnb_df["month_date"].head()
```

```python
# Check the first and last month available in the Airbnb dataset
print("First month:", airbnb_df["month_date"].min())

print("Last month:", airbnb_df["month_date"].max())
```

### Dropping unnecesary columns

```python
airbnb_df.columns
```

```python
columns_to_drop = [
    "host_id",
    "country",
    "currency",
    "native_avg_daily_rate",
    "native_revenue",
    "native_revpar",
    "num_reviews",
    "star_rating",
    "superhost",
    "booked_rate_avg",
    "listing_type",
    "unavailable_days"
]

airbnb_df = airbnb_df.drop(columns=columns_to_drop)
```
